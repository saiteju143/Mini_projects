```python

class Product:
    def __init__(self,product_id,name,brand,price,stock):
        self.product_id=product_id
        self.name=name
        self.brand=brand
        self.price=price
        self.stock=stock
    def display_product(self):
        print("Product_id:" , self.product_id)
        print("Name:" , self.name)
        print("Price:" , self.price)
        print("Stock:" , self.stock)
    def update_stock(self,quantity):
        self.stock+=quantity

class Electronics(Product):
    def __init__(self,product_id,name,brand,price,stock,warranty,model):
        super().__init__(product_id,name,brand,price,stock)
        self.warranty=warranty
        self.model=model
       

class Clothing(Product):
    def __init__(self,product_id,name,brand,price,stock,size,material):
        super().__init__(product_id,name,brand,price,stock)
        self.size=size
        self.material=material

class Books(Product):
    def __init__(self,product_id,name,brand,price,stock,author,publisher):
        super().__init__(product_id,name,brand,price,stock)
        self.author=author
        self.publisher=publisher
```

```python
import pandas as pd
df=pd.read_csv("sales.csv")
```

```python
total_quantity_sold=df["quantity"].sum()
```

```python
df["revenue"]=df["price"]*df["quantity"]
```

```python
df
```

```python
average_transaction_value=df["revenue"].mean().round(2)
print(average_transaction_value)
```

```python
highest_value_transaction=df["revenue"].max()
print(highest_value_transaction)
```

```python
lowest_value_transaction=df["revenue"].min()
print(lowest_value_transaction)
```

```python
best_selling_product=df.groupby("product_name")["quantity"].sum().idxmax()
print(best_selling_product)
```

```python
lowest_selling_product=df.groupby("product_name")["quantity"].sum().idxmin()
print(lowest_selling_product)
```

```python
category_wise_sales=df["category"].value_counts()
print(category_wise_sales)
```

```python
category_wise_revenue=df.groupby("category")["revenue"].sum()
print(category_wise_revenue)
```

```python
product_wise_sales=df.groupby("product_name")["revenue"].sum()
print(product_wise_sales)
```

```python
city_wise_sales=df.groupby("customer_city")["revenue"].sum()
print(city_wise_sales)
```

```python
payment_method=df["payment_method"].value_counts()
print(payment_method)
```

```python
product_sales = df.groupby("product_name")["quantity"].sum()

top_5_products = product_sales.sort_values(ascending=False).head(5)

print(top_5_products)
```

```python
product_sales = df.groupby("product_name")["quantity"].sum()

bottom_5_products = product_sales.sort_values().head(5)

print(bottom_5_products)
```
