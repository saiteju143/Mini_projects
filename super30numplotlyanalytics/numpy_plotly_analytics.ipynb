```python
import plotly.graph_objects as go
import plotly.express as ex
```

```python
months = np.array([
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
])
monthly_revenue = np.array([
    45000, 48000, 52000, 50000,
    55000, 60000, 58000, 62000,
    68000, 72000, 85000, 95000
])
orders = np.array([
    450, 470, 500, 490,
    530, 570, 550, 590,
    640, 680, 750, 820
])
customers = np.array([
    380, 400, 430, 420,
    460, 500, 490, 530,
    570, 610, 680, 740
])
expenses = np.array([
    30000, 31000, 33000, 32500,
    35000, 37000, 36000, 38000,
    40000, 42000, 47000, 52000
])
```

```python
total_annual_revenue=monthly_revenue.sum()
print(total_annual_revenue)
```

```python
average_monthly_revenue=monthly_revenue.mean()
print(average_monthly_revenue)
```

```python
monthly_revenue_max=monthly_revenue.argmax()
best_month=months[monthly_revenue_max]
print(best_month)
```

```python
monthly_revenue_min=monthly_revenue.argmin()
lowest_month=months[monthly_revenue_min]
print(lowest_month)
```

```python
month_to_month_revenue=np.diff(monthly_revenue)
print(month_to_month_revenue)
```

```python
fig=go.Figure()
fig.add_trace(go.Scatter(
    x=months,
    y=monthly_revenue,
   mode="lines+markers",
   name="Revenue"

))

fig.update_layout(
    title="Monthly_Revenue_Trend",
    xaxis_title="Months",
    yaxis_title="Revenue",
    showlegend=True
)
fig.show()
```

```python
fig=go.Figure()
fig.add_trace(
    go.Bar(
        x=months,
        y=orders,
        name="orders"
    )
)
fig.update_layout(
    title="Monthly_Orders",
    xaxis_title="Months",
    yaxis_title="Orders",
    showlegend=True

)

fig.show()
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=months,
        y=monthly_revenue,
        mode="lines+markers",
        name="Revenue"
    )
)

fig.add_trace(
    go.Scatter(
        x=months,
        y=expenses,
        mode="lines+markers",
        name="Expenses"
    )
)

fig.update_layout(
    title="Revenue vs Expenses",
    xaxis_title="Month",
    yaxis_title="Amount",
    showlegend=True
)

fig.show()
```

```python
fig=go.Figure()
fig.add_trace(go.Bar(
    x=months,
    y=customers,
    name="Customers_data"
))
fig.update_layout(
    title="Customer_Growth",
    xaxis_title="Months",
    yaxis_title="Customers",
    showlegend=True
)
fig.show()
```

```python
profit=monthly_revenue-expenses
fig=go.Figure()
fig.add_trace(
    go.Scatter(
        x=months,
        y=profit,
        mode="lines+markers",
        name="Profit"
    )
)

fig.update_layout(
    title="Profit Details",
    xaxis_title="Month",
    yaxis_title="Profit",
    showlegend=True
)

fig.show()
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=months,
        y=monthly_revenue,
        mode="lines+markers",
        name="Revenue"
    )
)

fig.add_trace(
    go.Scatter(
        x=months,
        y=expenses,
        mode="lines+markers",
        name="Expenses"
    )
)

fig.add_trace(
    go.Scatter(
        x=months,
        y=profit,
        mode="lines+markers",
        name="Profit"
    )
)

fig.update_layout(
    title="Overall Company Performance",
    xaxis_title="Month",
    yaxis_title="Amount",
    showlegend=True
)

fig.show()
```
