```python
with open("application.log" , "r") as file:
    data=file.read()
print(data)
```

```python
import pandas as pd
df=pd.read_csv("application.log",
               sep="-",
               header=None,
               names=["level" , "message"]
               )
print(df)
```

```python
len(df)
```

```python
df["level"].value_counts()
```

```python
with open("errors.txt" , "w") as file:
    for message in df.loc[df["level"]=="ERROR" , "message"]:
        file.write(message+"\n")
```

```python
with open("errors.txt" , "r") as file:
    for line in file:
        print(line,end="")
```

```python
with open("warning.txt" , "w") as file:
    for message in df.loc[df["level"]=="WARNING" , "message"]:
        file.write(message + "\n")
```

```python
with open("warning.txt" , "r") as file:
    for line in file:
        print(line,end="")
```
