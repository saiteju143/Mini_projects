```python
import plotly.graph_objects as go
import plotly.express as px
import numpy as np
```

```python
months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun"])
sales = np.array([120, 150, 170, 140, 200, 230])
```

```python
fig=go.Figure()
fig.add_trace(go.Bar(
               x=months,
               y=sales,
               name="Sales_Bar"
               ))
fig.update_layout(
    title="Monthly_Sales",
    xaxis_title="Months",
    yaxis_title="Sales",
    showlegend=True
)

fig.show()
```

```python
fig=go.Figure()
fig.add_trace(go.Scatter(
               x=months,
               y=sales,
               mode="lines+markers",
               name="Sales_Bar"
               ))
fig.update_layout(
    title="Monthly_Sales",
    xaxis_title="Months",
    yaxis_title="Sales",
    showlegend=True
)

fig.show()
```

```python
department=["Engineering " , "Sales" ,"Marketing" , "HR" , "Operations"]
Employees=[40,25,15,10,10]
fig=go.Figure()
fig.add_trace(go.Pie(
    labels=department,
    values=Employees
))
fig.update_layout(
    title=" department-wise employee distribution"
)
fig.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

laptop_sales = [120, 150, 180, 160, 210, 240]

mobile_sales = [200, 220, 250, 230, 270, 300]

tablet_sales = [80, 90, 110, 100, 130, 150]
fig=go.Figure()
fig.add_trace(go.Bar(
    x=months,
    y=laptop_sales,
    name="Laptop"
))
fig.add_trace(go.Bar(
    x=months,
    y=mobile_sales,
    name="Mobile_Sales"
))
fig.add_trace(go.Bar(
    x=months,
    y=tablet_sales,
    name="Tablet_Sales"
))
fig.update_layout(
    title="Monthly_Sales_Comparison",
    xaxis_title="Months",
    yaxis_title="Sales",
    barmode="group"
)
fig.show()
```
