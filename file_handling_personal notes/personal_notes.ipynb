```python
with open("notes.txt" , "w") as file:
    file.write("Python is easy to learn.\n")
    file.write("I am learning file handling.\n")
    file.write("Files are used to store data.\n")
    file.write("The open function opens a file.\n")
    file.write("The write function adds data.\n")
    file.write("The read function reads data.\n")
    file.write("The readline function reads one line.\n")
    file.write("The readlines function reads all lines.\n")
    file.write("The append mode adds new data.\n")
    file.write("I enjoy learning Python \n")
```

```python
with open("notes.txt" , "r") as file:
    file.read()
```
